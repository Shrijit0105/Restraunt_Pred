```python
df.isnull().sum()

if df.isnull().sum().any() > 0:
    print("There are missing values in the dataset.")

else:
    print("There are no missing values in the dataset.")


    
```

```python
print("The missing values are in the following columns:")
print(df[df.isnull().any(axis=1)])
```

```python
duplicate_rows = df.duplicated().sum()
if duplicate_rows > 0:
    print("There are duplicate rows in the dataset.")
    print(duplicate_rows)
else:
    print(f"There are no duplicate rows in the dataset.",duplicate_rows)    
```

```python
print("the unique values in the dataset are:")
if df.nunique().any() > 0:
    print(df.nunique())
else:
    print("There are no unique values in the dataset.")    
```

```python
print("The summary statistics for numeric columns are:")
print(df.describe(),"\n")
print("The summary statistics of all the columns in the dataset are:")
print(df.describe(include='all'),"\n")
```

```python
#Memory Usage of the DataFrame

print("The memory usage of the DataFrame is:")
print(df.memory_usage(deep=True),"\n")
print("The memory usage of the DataFrame is:(in KB)")
print(f"{df.memory_usage(deep=True).sum()/1024:.2f} KB","\n")
```
