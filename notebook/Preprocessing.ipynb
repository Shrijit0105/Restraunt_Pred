```python
# Feature Engineering
import sklearn
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer

df['Cuisines_Count']=df['Cuisines'].apply(lambda x: x.split(','))

mlb = MultiLabelBinarizer()

cuisines_encoded = mlb.fit_transform(df['Cuisines'].str.split(','))

cuisines_df = pd.DataFrame(cuisines_encoded, columns=[f'Cuisine_{c}' for c in mlb.classes_], index=df.index)

df = pd.concat(
    [
        df.drop(columns=['Cuisines']),
        cuisines_df
    ],
    axis=1
)

```

```python
# Log Transformation of Votes
```

```python
import numpy as np
df['Log_Votes'] = np.log1p(df['Votes'])
```

```python
import numpy as np
df['Cost_Price_Level']=(df['Average Cost for two'] / df['Price range'].replace(0, np.nan))
df['Cost_Price_Level'] =(df['Cost_Price_Level'].fillna(0))
```

```python
# Encoding cat features
cat_cols={'Has Table booking',
'Has Online delivery',
'Is delivering now'
}
df['Has Table booking']= df['Has Table booking'].replace({'Yes': 1, 'No': 0})

df['Has Online delivery']= df['Has Online delivery'].replace({'Yes': 1, 'No': 0})

df['Is delivering now']= df['Is delivering now'].replace({'Yes': 1, 'No': 0})
```
